# Number Statistics

This project analyzes a manually created list of numerical values.

The program:
- Identifies even and odd numbers.
- Creates separate lists for even and odd numbers.
- Finds the largest number.
- Finds the smallest number.

Python loops, conditional statements, the modulo operator, `max()`, and `min()` are used for the analysis.

```python
# Manually created numerical data

numbers = [12, 7, 25, 4, 18, 9, 30, 15]

print("Original Numbers:", numbers)
```

```python
# Create separate lists for even and odd numbers

even_numbers = []
odd_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)
    else:
        odd_numbers.append(number)

print("Even Numbers:", even_numbers)
print("Odd Numbers:", odd_numbers)
```

```python
# Find the largest and smallest numbers

largest_number = max(numbers)
smallest_number = min(numbers)

print("Largest Number:", largest_number)
print("Smallest Number:", smallest_number)
```

```python
# Final Number Statistics

print("===== Number Statistics =====")
print("Original Numbers:", numbers)
print("Even Numbers:", even_numbers)
print("Odd Numbers:", odd_numbers)
print("Largest Number:", largest_number)
print("Smallest Number:", smallest_number)
```
